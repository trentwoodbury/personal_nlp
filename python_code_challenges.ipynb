The Scenario: You are working for an e-commerce platform. You have a stream of raw transaction data. To prevent server crashes, you need to write a function that identifies "Flash Sale" spikes in real-time.The Data: A list of tuples representing (timestamp, transaction_amount). The timestamps are integers (seconds since epoch) and are not guaranteed to be perfectly spaced (irregular intervals).
<br>
The Task: Write a Python function detect_spikes(data, window_size, threshold) that: Calculates a Rolling Mean of the transaction_amount over a moving time window (e.g., the last 60 seconds).Identifies any data point where the transaction_amount is greater than threshold * rolling_mean.The Catch: You cannot use pandas. You must implement the rolling window logic using a standard library (hint: collections.deque is your friend for $O(1)$ efficiency).

### Input Example: (timestamp, amount)
transactions = [
    (100, 10), (102, 12), (105, 11), (110, 50), (120, 15), (180, 20)
]

window_size = 30 (seconds), threshold = 2.0

```python
import numpy as np
```

```python
def detect_spikes(data: list, window_size: int, threshold: float):
    """
    Returns all timestamps where the rolling average has exceeded 
    threshold * rolling mean.
    """
    timestamps = np.array([d[0] for d in data])
    amounts = np.array([d[1] for d in data])
    spikes = []

    for evaluation_idx in range(1, timestamps.shape[0]):
        current_timestamp, current_amount = timestamps[evaluation_idx], amounts[evaluation_idx]
        window_start = current_timestamp - window_size
        comparison_window_filter = [timestamp > window_start  and timestamp < current_timestamp for timestamp in timestamps]
        comparison_amounts = amounts[comparison_window_filter]
        amount_mean = np.mean(comparison_amounts)
        if current_amount > threshold * amount_mean:
            spikes.append((current_timestamp, current_amount))
    return spikes

```
